### Node Structure
class Node:
    def __init__(self,x):
        self.data=x
        self.next=None
#Stack implementing using Linked List
class myStack:
    def __init__(self):
        self.top=None
        self.count=0
    #push operation
    def push(self,x):
        temp=Node(x)
        temp.next=self.top
        self.top=self.top.next
        self.count==1
        return temp.data
    #peek operation
    def peek(self):
        if self.top is None:
            print("Stack is Empty")
            return -1
        return self.top.data
    #check if the stack is Empty
    def isEmpty(self):
        return self.top is None
    #size of stack
    def size(self):
        return self.count
#Main program
st=myStack()
n=int(input("Enter the no. of. elements to push:"))
for i in range(n):
    x=int(input("Enter the no of elements to push:"))
    st.push(x)
print("\n Popped element:",st.pop())
print("Top element:",st.peek())
if st.isEmpty():
    print("Stack is Empty")
else:
    print("stack is not empty")
print("Current Size:",st.size())
