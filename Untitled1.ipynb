```python
a=int(input("Enter number1:"))
b=int(input("Enter number2:"))

print("Enter the Sno to perform the Operations given below...")
print("1. Addition")
print("2. Subtraction")
print("3. Multiplication")
print("4. Division")
n=int(input("Enter the number of Operation to be performed:"))
print("A=",a)
print("B=",b)
switch(n)
case1: print("Addition:",a+b)
break
case2:print("Subtraction:",a-b)
break
case3:print("Multiplication:",a*b)
break
case4:print("Division:",a/b)
```
