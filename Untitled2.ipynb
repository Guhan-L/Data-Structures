```python
a= int(input("Enter number1: "))
b = int(input("Enter number2: "))
print("1. Addition")
print("2. Subtraction")
print("3. Multiplication")
print("4. Division")
n = int(input("Enter the number of Operation to be performed: "))
print("A =", a)
print("B =", b)
if n == 1:
    print("Addition:", a + b)
elif n == 2:
    print("Subtraction:", a - b)
elif n == 3:
    print("Multiplication:", a * b)
elif n == 4:
    print("Division:",a/b)
else:
    print("Invalid Operation selected.")

```
