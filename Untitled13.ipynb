```python
import matplotlib.pyplot as plt
import pandas as pd
df = pd.read_csv("Book2cia exam.csv")
df.head()
df.isnull().sum()
plt.xlabel("height")
plt.ylabel("weight")
plt.title("height and weight")
plt.scatter(df["height"],df["weight"],color="blue",label="line-1",marker="D",s=120)
plt.legend()
plt.grid(True)
plt.show()
```
